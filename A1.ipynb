# Part 3

```python
#!pip install torchmetrics
from torchmetrics.classification import BinaryJaccardIndex
```

```python
test_dataset = ImageDataset(0.8, 0.9, device="cuda:2")
```

```python
testX, testy = zip(*test_dataset.labeled_data)
testX = torch.stack(list(testX), 0)
testy = torch.stack(list(testy), 0)
```

```python
jaccard = BinaryJaccardIndex().to(device="cuda:2")
```

## First model - Very Simple

```python
simpmodel.eval()
```

```python
with torch.no_grad():
    simpoutput = simpmodel(testX)
```

```python
jaccard(simpoutput.squeeze(), testy)
```

## Second model - UNet

```python
UNetmodel.eval()
```

```python
with torch.no_grad():
    UNetoutput = UNetmodel(testX)
```

```python
jaccard(UNetoutput.squeeze(), testy)
```

## Plotting
